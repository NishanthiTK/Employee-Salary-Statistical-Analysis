## Skewness Analysis

```python
sample_std = np.std(salary, ddof=1)

print("Sample Standard Deviation:", sample_std)
```

```python
n = len(salary)
mean_salary = np.mean(salary)

skewness = (
    n / ((n - 1) * (n - 2))
) * np.sum(((salary - mean_salary) / sample_std) ** 3)

print("Skewness:", skewness)
```

```python
if skewness > 0.5:
    print("The salary distribution is positively skewed (Right-Skewed).")

elif skewness < -0.5:
    print("The salary distribution is negatively skewed (Left-Skewed).")

else:
    print("The salary distribution is approximately symmetric.")
```

## One-Way ANOVA

```python
salary_groups = []

for dept in unique_departments:
    dept_salary = salary[department == dept]
    salary_groups.append(dept_salary)

print("Number of Departments:", len(salary_groups))
```

```python
grand_mean = np.mean(salary)

print("Overall Mean Salary:", grand_mean)
```

```python
between_group = 0

for dept in unique_departments:
    
    dept_salary = salary[department == dept]
    
    dept_mean = np.mean(dept_salary)
    dept_count = len(dept_salary)
    
    between_group += dept_count * (dept_mean - grand_mean) ** 2

print("Between-Group Sum of Squares:", between_group)
```

```python
within_group = 0

for dept in unique_departments:
    
    dept_salary = salary[department == dept]
    dept_mean = np.mean(dept_salary)
    
    within_group += np.sum(
        (dept_salary - dept_mean) ** 2
    )

print("Within-Group Sum of Squares:", within_group)
```

```python
k = len(unique_departments)
n = len(salary)

df_between = k - 1
df_within = n - k

print("Number of Groups:", k)
print("Total Employees:", n)
print("Between-Group Degrees of Freedom:", df_between)
print("Within-Group Degrees of Freedom:", df_within)
```

```python
MS_between = between_group / df_between
MS_within = within_group / df_within

print("Mean Square Between:", MS_between)
print("Mean Square Within:", MS_within)
```

```python
F_value = MS_between / MS_within

print("F-Statistic:", F_value)
```

### Calculate p-value

```python
from scipy.stats import f

p_value = f.sf(
    F_value,
    df_between,
    df_within
)

print("p-value:", p_value)
```

```python
alpha = 0.05

print("Significance Level:", alpha)
```

```python
if p_value < alpha:
    print("Reject the Null Hypothesis.")
    print("There is a statistically significant difference in salary between departments.")

else:
    print("Fail to Reject the Null Hypothesis.")
    print("There is no statistically significant difference in salary between departments.")
```

```python
print("========== ONE-WAY ANOVA ==========")

print("Number of Departments:", k)
print("Total Employees:", n)
print("Grand Mean Salary:", round(grand_mean, 2))

print("Between-Group SS:", round(between_group, 2))
print("Within-Group SS:", round(within_group, 2))

print("Between-Group DF:", df_between)
print("Within-Group DF:", df_within)

print("MS Between:", round(MS_between, 2))
print("MS Within:", round(MS_within, 2))

print("F-Statistic:", round(F_value, 4))
print("p-value:", p_value)
```

## Chi-Square

```python
from scipy.stats import chi2_contingency

# Get unique categories
genders = np.unique(gender)
departments = np.unique(department)

# Create contingency table
contingency_table = np.zeros(
    (len(genders), len(departments)),
    dtype=int
)

for i, g in enumerate(genders):
    for j, d in enumerate(departments):
        contingency_table[i, j] = np.sum(
            (gender == g) & (department == d)
        )

print("Contingency Table:")
print(contingency_table)
```

```python
chi2_stat, p_value, dof, expected = chi2_contingency(
    contingency_table
)

print("Chi-Square Statistic:", chi2_stat)
print("p-value:", p_value)
print("Degrees of Freedom:", dof)

print("\nExpected Frequencies:")
print(expected)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis.")
    print("Gender and Department are significantly associated.")
else:
    print("Fail to Reject the Null Hypothesis.")
    print("Gender and Department are independent.")
```

```python
print("========== CHI-SQUARE TEST ==========")
print("Variables: Gender vs Department")
print("Chi-Square Statistic:", round(chi2_stat, 4))
print("Degrees of Freedom:", dof)
print("p-value:", p_value)
print("Significance Level:", alpha)
```
